```python
Fruits= ["Apple","Banana","Orange","Mango"]
print(Fruits)
print(Fruits[0])
print(Fruits[1])
print(Fruits[2])
print(Fruits[3])
print(Fruits[0:2])
print(Fruits[-1])
print(Fruits[:-3])
```

```python
Fruits[1]="Papaya"
print(Fruits)
```

```python
Fruits.append("Grapes")
print(Fruits)
```

```python
Fruits.extend(["Carrot","pineapple"])
print(Fruits)
```

```python
Fruits.remove("Orange")
print(Fruits)
```

```python
del Fruits[1]
print(Fruits)
```

```python
Fruits.clear()
print(Fruits)
```

```python
numbers=[10,78,34,87,90]
numbers.sort()
print(numbers)
numbers.sort(reverse=True)
print(numbers)
```

```python
colours=("red","green","blue")
print(colours)
print(type(colours))
```

```python
print(colours[0])
print(colours[1])
print(colours[-1])
```

```python
num=(10,20,30,40,50,60,70,80)
print(num[1:3])
print(num[:3])
print(num[3:5])
```

```python
# we cant edit the tuples but we can creat the  new tuple using the previous one
numb = num[:3] + (40) + num[4:]
print(numb)
```

```python
num=num + (90,)
print(num)
```

```python
numberss=(12,45,67,89,76,56)
result=tuple(sorted(numberss))
print(result)
```
