```python
fruits={"Mango","Mango","Peach","Carrot","Orange"}
unique_value=set(fruits)
print(unique_value)
print(len(unique_value))
```

```python
fruits.add("Banana")
print(fruits)
```

```python
fruits.remove("Banana")
print(fruits)
```

```python
setA={12,56,78,89,56,90,87}
setB={34,67,45,90,87,12}
print(setA & setB)
print(setA | setB)
print(setA -setB)
```
