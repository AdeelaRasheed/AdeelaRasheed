```python
sales_data=[1500,3000,6000,8500,12000]
for sales in sales_data:
  if sales<2000:
    category="Low Sales"
  elif sales<=5000:
    category="Medium sales"
  else:
    category="High sales"
  print(sales, "=", category)
```
