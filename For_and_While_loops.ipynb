```python
#use of for loop
for i in range(1,6):
  print(i)
```

```python
for i in range(2, 21, 2):
  print(i)
```

```python
#use of while loop
i=20
while i<= 30:
  print(i)
  i=i+1
```

```python
i=10
while i>=1:
  print(i)
  i=i-1
```

```python
students=["Ali","Mouhammad","Ahmed","sara"]
for students in students :
  print(students,  "- present")
```

```python
correct_pin = 1234
attempt = 0
while attempt < 3:
  pin = int(input("enter pin: "))
  if pin == correct_pin:
    print("login successful")
    break
  else:
    print("Wrong Pin")
    attempt += 1
```

```python
#List comprehension
numbers = [i for i in range (1,20)]
print(numbers)
```

```python
number2 = [i for i in range (2,21,2)]
print(number2)
```

```python
sales = [1000, 2500, 4000, 5000, 3000]
high_sales = [sale for sale in sales if sale > 2500]
print(high_sales)
```

```python
numbers3 = [i for i in range(1,21,2)]
print(numbers3)
```

```python
marks = [100,56,98,67,88,76,55,34,90]
high_marks = [mark for mark in marks if mark >50]
print(high_marks)
```

```python
numbers4 = [2,4,6,8,10,12]
multiple = [num * 2 for num in numbers4]
print(multiple)
```
